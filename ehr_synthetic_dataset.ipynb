# Creation of Medical Notes Dataset

```python
import random
from datasets import load_dataset

dataset = load_dataset("AGBonnet/augmented-clinical-notes")

sampled_data = dataset["train"].shuffle(seed=42).select(range(500))

sampled_data = sampled_data.remove_columns(["idx"])

patient_ids = [f"p{i+1:03d}" for i in range(len(sampled_data))]
sampled_data = sampled_data.add_column("patient_id", patient_ids)

sampled_data.to_csv("medical_notes_dataset.csv")

print("New dataset created with 500 random rows. Saved as 'medical_notes_dataset.csv'.")
```
