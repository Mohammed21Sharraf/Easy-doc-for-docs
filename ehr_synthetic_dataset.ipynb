```python
import random
import pandas as pd

patients = ["P{:03d}".format(i) for i in range(1, 501)]
symptoms_pool = [
    "fever, headache", "cough, sore throat", "stomach pain, diarrhea", "rash, itching",
    "fatigue, dizziness", "chest pain, shortness of breath"
]
diagnosis_pool = ["Dengue fever", "Common cold", "Food poisoning", "Allergy", "Hypertension", "Diabetes", "Asthma", 'COPD', 'Heart Failure']
prescriptions_pool = ["Paracetamol", "Cough syrup", "ORS", "Antibiotics", "Vitamin C", 'Metformin', 'Lisinopril', 'Albuterol', 'Prednisone', 'Digoxin']
doctor_notes_pool = [
    "Patient shows improvement but advised rest.",
    "Symptoms mild. Hydration recommended.",
    "Needs follow-up consultation.",
    "Prescribed bed rest for recovery.",
    "Symptoms severe, referred to specialist."
]
procedures_pool = [
    "Blood transfusion", "Appendectomy", "Physical therapy", 
    "Angioplasty", "Vaccination", "Wound dressing"
]
tests_pool = [
    "CBC (Complete Blood Count)", "X-ray", "MRI scan", 
    "Urine test", "Blood sugar test", "Lipid profile"
]
languages = ["English", "Bengali"]
sex_pool = ["Male", "Female"]
age_range = (1, 80)

data = []
for patient in patients:
    record = {
        "PatientID": patient,
        "VisitDate": f"2025-{random.randint(1, 12):02d}-{random.randint(1, 28):02d}",
        "Symptoms": random.choice(symptoms_pool),
        "Diagnosis": random.choice(diagnosis_pool),
        "Prescriptions": ", ".join(random.sample(prescriptions_pool, 2)),
        "Procedures": ", ".join(random.sample(procedures_pool, 2)),
        "Tests": ", ".join(random.sample(tests_pool, 2)),
        "DoctorNotes": random.choice(doctor_notes_pool),
        "Language": random.choice(languages)
    }
    data.append(record)

# Convert to a DataFrame
df = pd.DataFrame(data)

df.to_csv("synthetic_ehr_dataset.csv", index=False)

print("Dataset created and saved as 'synthetic_ehr_dataset.csv'")
```
